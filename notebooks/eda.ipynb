```python
# Data handling
import pandas as pd
import numpy as np

# Visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Settings
plt.style.use('ggplot')
sns.set(font_scale=1.1)

# Ignore warnings
import warnings
warnings.filterwarnings('ignore')
```

```python
# Load dataset
df = pd.read_csv('../data/raw/tcc_ceds_music.csv')

# Show first 5 rows
df.head()
```

```python
# Shape of dataset
print("Dataset Shape:")
print(df.shape)

# Column names
print("\nColumns:")
print(df.columns)

# Data types
print("\nData Types:")
print(df.dtypes)

# Basic statistics
print("\nStatistics:")
df.describe()
```

```python
# Missing values
missing = df.isnull().sum()

print(missing)
```

```python
plt.figure(figsize=(12,6))

sns.heatmap(df.isnull(), cbar=False)

plt.title("Missing Values Heatmap")
plt.show()
```

```python
# Check duplicates
print("Duplicate Rows:", df.duplicated().sum())

# Remove duplicates
df = df.drop_duplicates()
```

```python
text_columns = ['genre', 'artist_name', 'track_name']

for col in text_columns:
    df[col] = df[col].fillna('Unknown')
```

```python
df.isnull().sum()
```

```python
# Top genres
top_genres = df['genre'].value_counts().head(10)

print(top_genres)
```

```python
plt.figure(figsize=(12,6))

sns.countplot(
    y='genre',
    data=df,
    order=df['genre'].value_counts().index[:10]
)

plt.title("Top 10 Genres")
plt.xlabel("Count")
plt.ylabel("Genre")

plt.show()
```

```python
top_artists = df['artist_name'].value_counts().head(10)

print(top_artists)
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    x=top_artists.values,
    y=top_artists.index
)

plt.title("Top 10 Artists")
plt.xlabel("Song Count")
plt.ylabel("Artist")

plt.show()
```

```python
# Select numeric columns
numeric_df = df.select_dtypes(include=np.number)

# Correlation matrix
corr = numeric_df.corr()

plt.figure(figsize=(14,10))

sns.heatmap(
    corr,
    cmap='coolwarm',
    annot=False
)

plt.title("Feature Correlation Heatmap")

plt.show()
```

```python
df.to_csv('../data/processed/cleaned_music_dataset.csv', index=False)

print("Cleaned dataset saved successfully!")
```
